# 06 探索: Searching

## 課題
線形探索は、リストの先頭から順番に要素を調べていく。講義で示したコード`linear_search()`に、対象との比較回数を数える変数`count`を追加して、探索の効率を評価できるようにしなさい。
また、そのコードを用いて、長さ`n`のランダムな整数のリストを生成し、対象の整数を探索し、比較回数を出力するしなさい。
試行を$m$回行い、平均比較回数を計算して出力しなさい。

The linear search checks each element in order. 
Add a variable `count` to the code of `linear_search()` shown in the lecture to count the number of comparisons made, allowing you to evaluate the efficiency of the search.
Then, using that code, generate a list of random integers of length `n`, search for a target integer, and output the number of comparisons made.

## 課題
前問と同様に、二分探索のコード`binary_search()`に、対象との比較回数を数える変数`count`を追加して、探索の効率を評価できるようにしなさい。
そのコードを用いて、長さ`n`のランダムな整数のリストを生成し、対象の整数を探索し、比較回数を出力するしなさい。
試行を$m$回行い、平均比較回数を計算して出力しなさい。
さらに、$n$を2倍、4倍、8倍と増やしていったときの平均比較回数の変化を観察しなさい。

Similar to the previous question, add a variable `count` to the code of `binary_search()` to count the number of comparisons made, allowing you to evaluate the efficiency of the search.
Then, using that code, generate a list of random integers of length `n`, search for a target integer, and output the number of comparisons made.
Perform $m$ trials and calculate and output the average number of comparisons.
Additionally, observe how the average number of comparisons changes as $n$ is doubled, quadrupled, and octupled.
